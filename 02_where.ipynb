```python
%sql
SELECT * FROM employees
WHERE salary>50000;

SELECT * FROM employees
WHERE city="Delhi";

SELECT * FROM employees
WHERE joining_date>"2022-12-31";

SELECT * FROM employees
WHERE YEAR(joining_date)>'2022';


SELECT * FROM employees
WHERE MONTH(joining_date)>'05'; 


SELECT * FROM employees
WHERE DAY(joining_date)>'22';

SELECT * FROM employees
WHERE department_id !=2;

SELECT * FROM employees
WHERE salary BETWEEN 50000 AND 1000000;

SELECT * FROM employees
WHERE city IN("Delhi","Gurgaon");

SELECT * FROM employees
WHERE city NOT IN("Delhi");

SELECT * FROM employees
WHERE city="Delhi" OR city="Gurgaon";

SELECT * FROM employees
WHERE employee_name LIKE 'A%';

SELECT * FROM employees
WHERE employee_name LIKE '%a';

SELECT * FROM employees
WHERE employee_name LIKE '%a%';
```
