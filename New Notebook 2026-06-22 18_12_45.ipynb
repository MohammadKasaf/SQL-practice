Assign a unique row number to each employee based on sales amount in descending order.

```python
%sql
SELECT *,
ROW_NUMBER() OVER(ORDER BY sales_amount DESC) As row_num
FROM employee_sales;
```

Find the rank of employees based on sales amount.

```python
%sql
SELECT *,
RANK() OVER(ORDER BY sales_amount DESC) as rnk
FROM employee_sales;
```

Find employee ranking without rank gaps.

```python
%sql
SELECT *,
DENSE_RANK() OVER(ORDER BY sales_amount DESC) as rnk
FROM employee_sales;
```

Find top 2 highest sales employees from each department.

```python
%sql
SELECT * 
FROM (
SELECT *,
ROW_NUMBER() OVER(PARTITION BY department ORDER BY sales_amount DESC) AS row_number
FROM employee_sales
)t
WHERE row_number <=2;

```

Find employees having highest sales in every department.

```python
%sql

SELECT *
FROM (
SELECT *,
ROW_NUMBER()  OVER(PARTITION BY department ORDER BY sales_amount DESC) AS row_number
FROM employee_sales
)t
WHERE row_number=1;
```

Find previous employee sales amount for each employee ordered by sales.

```python
%sql
SELECT *,
LAG(sales_amount) OVER(ORDER BY sales_amount DESC) AS previous_sale
FROM employee_sales;
```

Find next employee sales amount for each employee.

```python
%sql
SELECT *,
LEAD(sales_amount) OVER(ORDER BY sales_amount DESC) AS previous_sale
FROM employee_sales;
```

Calculate sales difference from previous employee.

```python
%sql
SELECT *,
LAG(sales_amount) OVER(ORDER BY sales_amount DESC) AS prev_sale,(sales_amount-prev_sale) AS difference
FROM employee_sales;
```
