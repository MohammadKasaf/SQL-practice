```python
%sql
SELECT department,COUNT(emp_id)
FROM employees
GROUP BY department HAVING COUNT(emp_id)>3;
```

```python
%sql
SELECT department,AVG(salary)
FROM employees
GROUP BY department HAVING AVG(salary)>80000
```

```python
%sql
SELECT city,SUM(salary)
FROM employees
GROUP BY city HAVING SUM(salary)>200000
```

```python
%sql
SELECT department,MAX(salary),MIN(salary)
FROM employees
GROUP BY department HAVING MAX(salary)>90000 AND  MIN(salary)>50000
```

```python
%sql
SELECT department,COUNT(emp_id)
FROM employees
WHERE experience_years>5
GROUP BY department HAVING COUNT(emp_id)>1
```
