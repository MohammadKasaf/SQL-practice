Find employees whose salary is greater than the company average salary.


```python
%sql
SELECT * FROM employees
WHERE salary>(
    SELECT AVG(salary) FROM employees
    );
```

Second highest salary

```python
%sql
SELECT MAX(Salary) FROM employees
WHERE salary<(

    SELECT MAX(salary) FROM employees
);
```

Find all employees working in the department that has the highest average salary.

```python
%sql
   SELECT * FROM employees
   WHERE department=(
     SELECT department
     FROM employees
     GROUP BY department
     ORDER BY AVG(salary) DESC LIMIT 1 
    )
```

Find employees who earn the highest salary in their department.

```python
%sql
SELECT * FROM employees e1
WHERE salary=(

    SELECT MAX(salary) FROM employees e2
    WHERE e1.department=e2.department
);
```

Find employees whose salary is greater than ALL employees in the IT department.

```python
%sql
SELECT DISTINCT * FROM employees
WHERE salary >
  (

    SELECT MAX(salary)
    FROM employees
    WHERE department="IT"
);
```

Find customers who placed at least one order.

```python
%sql

SELECT * FROM employees
WHERE emp_name IN(
SELECT customer_name 
FROM orders
);
```

Find employees who never placed any order.

```python
%sql

SELECT * FROM employees
WHERE emp_name NOT IN(
SELECT customer_name 
FROM orders
);
```
